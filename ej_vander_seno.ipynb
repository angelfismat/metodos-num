```python
import numpy as np
#from scipy.interpolate import vander
from scipy import interpolate

import matplotlib.pyplot as plt
```

```python
x = np.linspace(-np.pi,np.pi,11)
y = np.sin(x)
```

```python
C=np.vander(x)
P=np.linalg.solve(C,y)
y1d = interpolate.interp1d(x, y)
```

```python
P
```

```python
x2 = np.linspace(-np.pi,np.pi,1001)
van = np.polyval(P, x2)


plt.scatter(x,y)
plt.plot(x2,y1d(x2),label=r"Lineal", linestyle='-')
plt.plot(x2,van,label=r"VanderMonde", linestyle='-.')
plt.plot(x2,np.sin(x2),label="Seno", linestyle='--')

plt.legend()
plt.grid()
plt.show()
```
