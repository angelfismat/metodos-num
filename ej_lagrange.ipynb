```python
import numpy as np
from scipy.interpolate import lagrange
from scipy import interpolate

import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1,1,11)
y = 1./(1+25*x**2)
```

```python
lag = lagrange(x, y)
y1d = interpolate.interp1d(x, y)
```

```python
lag
```

```python
x2 = np.linspace(-1,1,101)

plt.scatter(x,y)
plt.plot(x2,y1d(x2),label=r"Lineal", linestyle='-')
plt.plot(x2,lag(x2),label=r"Lagrange", linestyle='-.')

plt.legend()
plt.grid()
plt.show()
```
