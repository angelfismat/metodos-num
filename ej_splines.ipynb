```python
import numpy as np
from scipy.interpolate import lagrange
from scipy.interpolate import CubicSpline

import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1,1,11)
y = 1./(1+25*x**2)
```

```python
lag = lagrange(x, y)
y1s = CubicSpline(x, y)
```

```python
x2 = np.linspace(-1,1,101)

plt.scatter(x,y)
plt.plot(x2,y1s(x2),'b*',label=r"Splines", linestyle='--')
plt.plot(x2,lag(x2),label=r"Lagrange", linestyle='-.')
plt.plot(x2,1./(1+25*x2**2),label=r"real", linestyle=':')

plt.legend()
plt.grid()
plt.show()
```
