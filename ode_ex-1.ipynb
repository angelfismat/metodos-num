```python
import numpy as np
from scipy.optimize import fsolve
```

```python
def eulerf(a, b, n, f, y0):
    """Euler hacia adelante (Euler Explícito)"""
    x = np.linspace(a, b, n)  # Estas son las equises[cite: 2]
    h = x[1] - x[0]           # Tamaño de paso[cite: 2]
    
    y0 = np.atleast_1d(y0)
    y = np.zeros((len(y0), n))
    y[:, 0] = y0               # Inicializa[cite: 2]
    
    for i in range(1, n):
        y[:, i] = y[:, i-1] + h * np.array(f(x[i-1], y[:, i-1]))  # Euler hacia adelante[cite: 2]
        
    return y, x
```

```python
def eulerb(a, b, n, f, y0):
    """Euler hacia atrás (Euler Implícito)"""
    x = np.linspace(a, b, n)  # Estas son las equises[cite: 1]
    h = x[1] - x[0]           # Tamaño de paso[cite: 1]
    
    y0 = np.atleast_1d(y0)
    y = np.zeros((len(y0), n))
    y[:, 0] = y0               # Inicializa[cite: 1]
    
    for i in range(1, n):
        xtemp = x[i]
        ytemp = y[:, i-1]
        
        # Función implícita F(z) = 0
        F = lambda z: h * np.array(f(xtemp, z)) + ytemp - z[cite: 1]
        
        # fsolve busca la raíz z comenzando desde ytemp[cite: 1]
        y[:, i] = fsolve(F, ytemp)  # Euler hacia atrás[cite: 1]
        
    return y, x
```

```python
def trapos(a, b, n, f, y0):
    """Método de los Trapecios (Implícito)"""
    x = np.linspace(a, b, n)  # Estas son las equises[cite: 3]
    h = x[1] - x[0]           # Tamaño de paso[cite: 3]
    
    y0 = np.atleast_1d(y0)
    y = np.zeros((len(y0), n))
    y[:, 0] = y0               # Inicializa[cite: 3]
    
    for i in range(1, n):
        xtemp = x[i]
        ytemp = y[:, i-1]
        xtomp = x[i-1]
        
        # Función implícita F(z) = 0
        F = lambda z: h * (np.array(f(xtemp, z)) + np.array(f(xtomp, ytemp))) / 2 + ytemp - z[cite: 3]
        
        y[:, i] = fsolve(F, ytemp)  # Trapecios[cite: 3]
        
    return y, x
```

```python
# 1. Definición del sistema 2x2: dy/dt = f(t, y)
def sistema_2x2(t, y):
    y1, y2 = y[0], y[1]
    dy1_dt = y1 - 0.5 * y2
    dy2_dt = -0.2 * y1 + 0.8 * y2
    return [dy1_dt, dy2_dt]
```

```python
# 2. Parámetros de la simulación
a, b = 0.0, 2.0   # Intervalo de tiempo [t_inicial, t_final]
n = 50            # Número de puntos
y0 = [10.0, 5.0]  # Condiciones iniciales [y1(0), y2(0)]

# 3. Resolución con Euler hacia adelante
y, t = eulerf(a, b, n, sistema_2x2, y0)

# 4. Mostrar los primeros 5 resultados
print(f"{'Tiempo (t)':^12} | {'Especie 1 (y1)':^15} | {'Especie 2 (y2)':^15}")
print("-" * 48)
for i in range(5):
    print(f"{t[i]:^12.4f} | {y[0, i]:^15.4f} | {y[1, i]:^15.4f}")
```
